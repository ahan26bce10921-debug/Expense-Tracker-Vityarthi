```python
class Expense:
    def __init__(self, title, amount, category):
        self.title = title
        self.amount = float(amount)
        self.category = category



def display_menu():

    print("\n" + "=" * 30)
    print("   EXPENSE TRACKER SYSTEM")
    print("=" * 30)
    print("1. Add Expense")
    print("2. View All Expenses")
    print("3. View Total Spent")
    print("4. Exit")


def add_expense(expenses_list):

    title = input("Enter Expense Title: ")
    amount = float(input("Enter Amount (₹): "))
    category = input("Enter Category (Food/Rent/etc.): ")


    new_expense = Expense(title, amount, category)


    expenses_list.append(new_expense)
    print("\n--> Expense added successfully!")


def view_expenses(expenses_list):
    
    if not expenses_list:
        print("\n--> No expenses added yet.")
        return

    print("\n--- List of Expenses ---")
    print(f"{'S.No':<5} {'Title':<20} {'Category':<15} {'Amount (₹)':<10}")
    print("-" * 55)

    
    for index, exp in enumerate(expenses_list, start=1):
        print(
            f"{index:<5} {exp.title:<20} {exp.category:<15} ₹{exp.amount:<10.2f}"
        )


def show_total(expenses_list):
    total = 0.0
    
    for exp in expenses_list:
        total += exp.amount

    
    print(f"\n--> Total Money Spent: ₹{total:.2f}")


def main():
    
    expenses = []

    
    while True:
        display_menu()
        choice = input("Enter your choice (1-4): ")

        
        if choice == "1":
            add_expense(expenses)
        elif choice == "2":
            view_expenses(expenses)
        elif choice == "3":
            show_total(expenses)
        elif choice == "4":
            print("\nExiting program. Goodbye!")
            break
        else:
            print("\n--> Invalid choice! Please enter a number between 1 and 4.")



if __name__ == "__main__":
    main()
```
